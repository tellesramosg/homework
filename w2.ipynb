```python
name = "Giancarlo"
age = 18
grade = 100.0
is_student = True
favorite_color = "Blue"

print(name)
print(age)
print(grade)
print(is_student)
print(favorite_color)
```

```python
name: str = "Giancarlo"
age: int = 18
grade: float = 100.0
is_student: bool = True
favorite_color: str = "Blue"

name = name.upper()
print(name)
```

```python
age + grade # data type is a float since it is a decimal number

print(type(age + grade))
```

```python
person = {
    "name": name,
    "age": age,
    "grade": grade,
    "is_student": is_student,
    "favorite_color": favorite_color
}

print(person)
```

```python
print(f"My name is {person["name"]}. "
      f"I am {person["age"]} years old. "
      f"My favorite color is {person["favorite_color"]}. "
      f"I will get a {person["grade"]} in this class.")
```
