```python
import pandas as pd
```

```python
df = pd.read_csv("./data/french-ambs.csv")
df.head()
```

```python
df.shape

print("Number of Rows:", df.shape[0])
```

```python
filtered_ambs = df.loc[(df["year"] > 1588) & (df["year"] < 1592)]

print(filtered_ambs.shape[0])
print(filtered_ambs["name"].nunique())
```

```python
df["is_permenant"] = True
df.loc[df["place"].isin(["Geneva", "Saxony", "Tuscany", "Ferrara"]), "is_permenant"] = False
len(df.loc[df["is_permenant"] == False])
```

```python
df["place"] = df["place"].replace({"Geneva": "Swiss Cantons", "Grisons": "Swiss Cantons"})
len(df.loc[df["place"] == "Swiss Cantons"])
```

```python
df_agg = df.groupby("place").agg(
    years_present = ("place", "count"),
    unique_ambs = ("name", "nunique")
).sort_values("unique_ambs", ascending=True).reset_index()

df_agg
```
