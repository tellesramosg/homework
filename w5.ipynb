```python
import pandas as pd
```

```python
data = {
    "name": ["Nate", "John", "Jane", "Manday"],
    "age": [39, 23, 31, 25],
    "sex": ["m", "m", "f", "f"],
    "email": ["nate@gmail.com", "john@gmail.com", "jane@gmail.com", "mandy@gmail.com"],
    "purchases": [34, 12, 42, 8],
}

df = pd.DataFrame(data=data)
df
```

```python
df = df.rename(columns={"sex": "gender"})
df
```

```python
df["active"] = True
df
```
